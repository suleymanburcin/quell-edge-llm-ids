egitilebilecegini belirler. **GPU'lu dugumde** calistir.

Onaylanmis cikti (2026-07-22): 2x RTX A5500 (24 GB), 48 cekirdek, 123 GB RAM, CUDA 12.8.

```python
import subprocess, platform, os
def run(cmd):
    try: return subprocess.check_output(cmd, shell=True, stderr=subprocess.DEVNULL, text=True).strip()
    except: return None
print("OS:", platform.system(), platform.release())
print("CPU cores:", os.cpu_count())
ram = run("grep MemTotal /proc/meminfo")
print("RAM GB:", round(int(ram.split()[1])/1024**2,1) if ram else "?")
print("--- nvidia-smi ---")
print(run("nvidia-smi --query-gpu=name,memory.total,driver_version --format=csv,noheader") or "GPU YOK")
try:
    import torch
    print("torch:", torch.__version__, "| CUDA available:", torch.cuda.is_available(),
          "| CUDA:", torch.version.cuda, "| GPU count:", torch.cuda.device_count())
except Exception as e:
    print("torch yok ->", e)
```
